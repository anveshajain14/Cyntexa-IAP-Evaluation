# 2 what are the factors affecting sales in a region
Key Factors Affecting Regional Sales
Based on analysis of the TPC-H wholesale business data, regional sales are influenced by multiple interconnected factors. Despite Europe leading with $219.0B in revenue and Africa trailing at $216.7B, the regions are remarkably balanced—less than 1% difference—suggesting that no single factor dominates but rather a combination of factors creates the regional variations.
1. Customer Base Size and Density
Customer count is the strongest driver of regional revenue differences. Europe has the largest customer base with 100,317 customers, while Africa has the smallest with 99,622 customers—a ~700 customer difference that directly correlates with the revenue gap. 1
![image_1792666179739.png](./image_1792666179739.png "image_1792666179739.png")
2. Market Segment Composition
Market segment mix varies significantly by region and impacts total sales. In Asia, the Furniture segment generates $44.4B (the highest across all region-segment combinations), while in America, Furniture only generates $43.1B. Each region has different segment strengths:
Europe: Strong in Automobile ($44.2B) and Building ($44.1B)
Asia: Dominates in Furniture ($44.4B)
Middle East: Balanced across Furniture ($44.0B) and Automobile ($43.8B)
America & Africa: More evenly distributed across all segments
